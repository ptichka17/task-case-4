```python
def max_dragon_strength(N):
    if N <= 4:
        return N  # Для N=1,2,3,4 оптимально использовать N как одно число
    if N == 5:
        return 6  # 3 * 2 = 6
    if N == 7:
        return 12  # 3 * 4 = 12
    
    k = N // 3  # Количество драконов с 3 головами
    r = N % 3   # Остаток
    
    if r == 0:
        return 3 ** k
    elif r == 1:
        return 3 ** (k - 1) * 4
    else:  # r == 2
        return 3 ** k * 2

# Считываем входные данные
N = int(input())
# Выводим результат
print(max_dragon_strength(N))
```
